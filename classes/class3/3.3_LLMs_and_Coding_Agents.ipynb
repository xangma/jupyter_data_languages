# 3.3 LLMs and Coding Agents

Practical guidance for collaborating with any AI coding assistant while keeping scientific rigor front and center.


## Prompt Patterns for Everyday Coding Tasks
- **Refactor a function**: "Refactor the provided `combine_fluxes` helper to remove duplicate loops, keep numerical stability, and summarize each change in one sentence."
- **Write docstrings**: "Draft a concise Numpy-style docstring for `fit_blackbody`, including parameters, returns, and one usage note pulled from the surrounding context."
- **Generate tests**: "Given the interface for `compute_lightcurve`, propose two edge-focused unit tests with concrete inputs/expected outputs and name the tests clearly."
- **Explain an error message**: "Explain this `TypeError` traceback in plain language, point to the offending line, and suggest two minimal fixes ranked by safety."
- **Suggest vectorization**: "Show how to vectorize the nested loop in the snippet using NumPy broadcasting, highlight memory implications, and provide a drop-in replacement."


## Verify, Don't Trust Checklist
- [ ] Run the full script/notebook after applying any AI-generated change.
- [ ] Inspect numerical outputs, plots, and logs for plausibility.
- [ ] Compare against a baseline result or regression dataset.
- [ ] Add or update assertions/plots that lock in the expected behavior.


## Demo: Quick Test Catches an LLM Slip
We'll take an LLM-supplied helper, immediately wrap it in a lightweight unit test, let the test fail loudly, and then patch the function.


```python
import math

def llm_surface_gravity(mass_kg: float, radius_m: float) -> float:
    """LLM attempt: compute surface gravity of a spherical body in m/s^2."""
    G = 6.674e-11
    # Intentional slip: forgetting that gravity scales with radius squared
    return (G * mass_kg) / radius_m
```

```python
def test_surface_gravity():
    earth_mass = 5.972e24  # kg
    earth_radius = 6.371e6  # m
    expected = 9.81
    got = llm_surface_gravity(earth_mass, earth_radius)
    assert math.isclose(got, expected, rel_tol=1e-2), f"Expected {expected} m/s^2 but got {got:.2f}"

try:
    test_surface_gravity()
except AssertionError as err:
    print("Quick test caught an issue:", err)
```

```python
def llm_surface_gravity(mass_kg: float, radius_m: float) -> float:
    """Corrected: include the radius squared term for Newtonian gravity."""
    G = 6.674e-11
    return (G * mass_kg) / (radius_m ** 2)
```

```python
test_surface_gravity()
print("Test now passes — ready to trust but verify future tweaks.")
```

## Ethics & Confidentiality
- Never paste proprietary or patient data into an LLM unless governance explicitly permits it.
- Strip or anonymize observational metadata (coordinates, identifiers) before sharing context.
- Document when AI-assisted code enters a project so reviewers know what to scrutinize.
- Attribute external ideas/code snippets and check licenses before integrating.
- Default to the most restrictive data-handling policy when uncertain; ask for clearance rather than forgiveness.
