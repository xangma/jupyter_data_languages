# Plotting uncertainty
In this example we will go over plotting uncertainties in various ways:
+ y errorbars
+ x errorbars
+ x and y errorbars (no covariance)
+ x and y error-ellipse (covariance)

## Packages being used
+ `matplotlib`: all the plotting
+ `pandas`: read in the data table
+ `numpy` and `scipy`: convert cov matrix to ellipse params

## Relevant documentation
+ `matplotlib`: https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.errorbar.html

```python
import pandas
import numpy as np
from matplotlib import pyplot as plt
from matplotlib.patches import Ellipse
import sys
from pathlib import Path
_REPO_ROOT = Path.cwd().resolve()
while _REPO_ROOT != _REPO_ROOT.parent and not (_REPO_ROOT / 'utils' / 'mpl_style.py').exists():
    _REPO_ROOT = _REPO_ROOT.parent
if str(_REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(_REPO_ROOT))
from utils import mpl_style
%matplotlib inline
plt.style.use(mpl_style.style1)
```

Our data contains $(x, y)$ positions with 1-$\sigma$ uncertainties and covariance values:

```python
from pathlib import Path
data_path = Path('data.csv')
if data_path.exists():
    t = pandas.read_csv(data_path)
else:
    rng = np.random.default_rng(314)
    n_points = 15
    x = np.linspace(20, 400, n_points) + rng.normal(0, 15, n_points)
    y = np.linspace(350, 650, n_points) + rng.normal(0, 25, n_points)
    sx = rng.uniform(3, 12, n_points)
    sy = rng.uniform(10, 65, n_points)
    pxy = rng.uniform(-0.9, 0.9, n_points)
    t = pandas.DataFrame({'x': x, 'y': y, 'sx': sx, 'sy': sy, 'pxy': pxy})
display(t)
```

**Note** the full covariance matrix for each data point is: 
$\left[ \begin{array}{ccc} \sigma_x^2 & \rho_{xy}\sigma_x \sigma_y \\ \rho_{xy}\sigma_x \sigma_y & \sigma_y^2 \end{array} \right]$

## y-uncertainties or x-uncertainties only
The most common type of data you will work with will only have (significant) uncertainties in one direction.  In this case it is very easy to plot using `errorbar`:

```python
plt.figure(1)
plt.errorbar(
    t.x,
    t.y,
    yerr=t.sy,
    ls='None',
    mfc='k',
    mec='k',
    ms=5,
    marker='s',
    ecolor='k'
)
plt.xlabel('x')
plt.ylabel('y')
plt.ylim(0, 700)
```

```python
plt.figure(2)
plt.errorbar(
    t.x,
    t.y,
    xerr=t.sx,
    ls='None',
    mfc='k',
    mec='k',
    ms=5,
    marker='s',
    ecolor='k'
)
plt.xlabel('x')
plt.ylabel('y')
plt.ylim(0, 700);
```

## Uncertainties in both x and y with no cov
If your data has no cov you can still use `errorbar`:

```python
plt.figure(3)
plt.errorbar(
    t.x,
    t.y,
    yerr=t.sy,
    xerr=t.sx,
    ls='None',
    mfc='k',
    mec='k',
    ms=5,
    marker='s',
    ecolor='k'
)
plt.xlabel('x')
plt.ylabel('y')
plt.ylim(0, 700);
```

## Uncertainties in both x and y with cov
If your data does have cov you should plot a 1-$\sigma$ ellipse around each point.  There is no built in function to do this, so we will have to write our own.  We will start by writing a function to turn a cov matrix into the parameters for an ellipse and draw it on a figure.

```python
def cov_to_ellipse(cov, pos, **kwargs):
    eigvec,eigval,V = np.linalg.svd(cov,full_matrices=False)
    # the angle the first eigenvector makes with the x-axis
    theta = np.degrees(np.arctan2(eigvec[1, 0], eigvec[0, 0]))
    # full width and height of ellipse, not radius
    # the eigenvalues are the variance along the eigenvectors
    width, height = 2 * np.sqrt(eigval)
    return Ellipse(xy=pos, width=width, height=height, angle=theta, **kwargs)

def plot_ellipse(t, ax=None, **kwargs):
    if ax is None:
        ax = plt.gca()
    for rdx, row in t.iterrows():
        cov = np.array(
            [[row.sx**2, row.pxy * row.sx * row.sy],
             [row.pxy * row.sx * row.sy, row.sy**2]]
        )
        ellip = cov_to_ellipse(cov, [row.x, row.y], **kwargs)
        ax.add_artist(ellip)
        
plt.figure(4)
plt.plot(
    t['x'],
    t['y'],
    's',
    mfc='k',
    mec='k',
    ms=5
)
plot_ellipse(
    t,
    lw=1.5,
    fc='none',
    ec='C0'
)
plt.xlabel('x')
plt.ylabel('y')
plt.ylim(0, 700)
plt.draw();
```

## Saving Figures
Use `plt.savefig` to store publication-quality graphics. Set a high `dpi` (e.g., 300) for PNG output and also write a PDF when you need vector graphics for journals. The snippet below makes sure an `images/` folder exists before saving both formats.

```python
from pathlib import Path

demo_fig, demo_ax = plt.subplots(figsize=(4, 3))
x_demo = np.linspace(0, 2 * np.pi, 200)
demo_ax.plot(x_demo, np.sin(x_demo), label='demo')
demo_ax.set_xlabel('x')
demo_ax.set_ylabel('sin(x)')
demo_ax.legend(loc='best')

search_root = Path.cwd().resolve()
images_dir = None
while True:
    candidate = search_root / 'images'
    if candidate.exists():
        images_dir = candidate
    if search_root == search_root.parent:
        break
    search_root = search_root.parent
if images_dir is None:
    images_dir = Path.cwd().resolve() / 'images'
images_dir.mkdir(parents=True, exist_ok=True)
svg_path = images_dir / 'class2_uncertainty_plotting_demo.svg'
pdf_path = images_dir / 'class2_uncertainty_plotting_demo.pdf'
demo_fig.savefig(svg_path)
demo_fig.savefig(pdf_path)
plt.close(demo_fig)
print(f"Saved {svg_path} and {pdf_path}")
```
